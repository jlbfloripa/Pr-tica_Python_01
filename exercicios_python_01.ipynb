# Operadores aritiméticos

+ Adição
- Subtração
* Multiplicação
/ Divisão
// Divisão inteira (sem o resto)
% Resto da divisão
** Potenciação

```python
nota1 = 7.5
nota2 = 8.0

print(nota1 + nota2)
print("A soma das notas é: ", nota1 + nota2)
```

```python
nota1 = 7.5
nota2 = 8.0

print((nota1 + nota2) / 2)
print("A média das notas é: ", (nota1 + nota2) / 2)
```

```python
# Média é a soma das notas dividida pelo número de notas.

nota1 = 8
nota2 = 6
nota3 = 9

media = (nota1 + nota2 + nota3) / 3
print("A média das notas é: ", media)
```

```python
# Se fizer fora dos parênteses, a divisão será feita apenas com a última nota, e não com a soma das três.

nota1 = 8
nota2 = 6
nota3 = 9

media = nota1 + nota2 + nota3 / 3
print(media)  # Resultado incorreto, pois a divisão é feita apenas com a última nota.
```

# Tipos de dados

```python
# Usando dtype() para verificar o tipo de dado de cada variável

idade = 45 # não está entre aspas, então é um inteiro
nota = 8.5
nome = "Ana"

print(type(idade))
print(type(nota))
print(type(nome))
```

# Conversão de tipos de dados = type casting

Função int() para converter uma variável para inteiro

Função float() para converter uma variável para ponto flutuante (decimal)

Função str() para converter uma variável para string (texto)

```python
# Função int() para converter uma string em um número inteiro
# IMPORTANTE: int() não arredonda, ele apenas converte a string para inteiro, descartando a parte decimal.

idade = "25" # se estiver entre aspas, é uma string
idade = int(idade)  # Convertendo para inteiro

print(idade) # Verificando o valor após a conversão
print(type(idade))  # Verificando o tipo de dado após a conversão
```

```python
# Função float() para converter uma string em um número decimal

idade = "25"

idade = float(idade)  # Convertendo para decimal

print(idade) # Verificando o valor após a conversão
print(type(idade))  # Verificando o tipo de dado após a conversão
```

```python
# Função str() para converter um número em uma string

idade = 25

idade = str(idade)  # Convertendo para string

print(idade)
print(type(idade))  # Verificando o tipo de dado após a conversão
```

```python
# Converter idade para inteiro
# Converter altura para decimal
# Converter respostas para inteiro
# Usar a função type() para verificar o tipo de dado de cada variável

idade = "32"
altura = "1.68"
respostas = "15"

print(type(idade))  # Antes da conversão
print(type(altura))
print(type(respostas)) 
```

```python
idade = "32"
altura = "1.68"
respostas = "15"

idade = int(idade)  # Convertendo para inteiro
altura = float(altura)  # Convertendo para decimal
respostas = int(respostas)  # Convertendo para inteiro

print(type(idade))  # Verificando o tipo de dado após a conversão
print(type(altura))
print(type(respostas))
```
